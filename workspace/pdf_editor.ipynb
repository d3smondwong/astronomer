```python
from pypdf import PdfReader, PdfWriter

def delete_pages(input_pdf, output_pdf, pages_to_delete):
    """
    Deletes specific pages from a PDF.
    :param input_pdf: Path to the source PDF.
    :param output_pdf: Path where the new PDF will be saved.
    :param pages_to_delete: List of page indices to remove (0-indexed).
    """
    reader = PdfReader(input_pdf)
    writer = PdfWriter()

    # Loop through all pages in the original PDF
    for index in range(len(reader.pages)):
        # Only add the page if it's NOT in our delete list
        if index not in pages_to_delete:
            writer.add_page(reader.pages[index])

    # Save the new PDF
    with open(output_pdf, "wb") as f:
        writer.write(f)

    print(f"Success! New PDF saved as: {output_pdf}")

# Example Usage:
# Note: PDF pages are 0-indexed. Page 1 is index 0, Page 2 is index 1, etc.
pages_to_remove = [0, 2]  # This will delete the 1st and 3rd pages
delete_pages("source.pdf", "cleaned_result.pdf", pages_to_remove)
```

```python
from pypdf import PdfReader, PdfWriter

def extract_pages(input_pdf, output_pdf, pages_to_keep):
    """
    Extracts specific pages from a PDF.
    :param input_pdf: Path to the source PDF.
    :param output_pdf: Path where the extracted pages will be saved.
    :param pages_to_keep: List of page indices to extract (0-indexed).
    """
    reader = PdfReader(input_pdf)
    writer = PdfWriter()

    # Loop through the list of pages you actually want
    for page_num in pages_to_keep:
        # Check to ensure the page index exists in the PDF
        if page_num < len(reader.pages):
            writer.add_page(reader.pages[page_num])
        else:
            print(f"Warning: Page {page_num} is out of range.")

    with open(output_pdf, "wb") as f:
        writer.write(f)

    print(f"Extraction complete: {output_pdf}")

# Example Usage:
# To extract the 2nd, 5th, and 10th pages:
pages_idx = [1, 4, 9]
extract_pages("source.pdf", "extracted_pages.pdf", pages_idx)
```
